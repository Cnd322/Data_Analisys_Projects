```python
print(df.columns)
```

### 2.2. Типы данных



```python
df['year_of_release'] = pd.to_numeric(df['year_of_release'], errors='coerce').astype('Int64')
print(df.info()) 
```

- В числовых столбцах могут встретиться строковые значения, например `unknown` или другие. Приводите такие столбцы к числовому типу данных, заменив строковые значения на пропуски.

```python
for column in ['na_sales', 'eu_sales', 'jp_sales', 'other_sales', 'critic_score', 'user_score']:
    df[column] = pd.to_numeric(df[column], errors='coerce')
    
print(df.info()) #Привожу столбцы с данными, подразумевающие работу с числами к числовому формату
```

### 2.3. Наличие пропусков в данных


```python
df.isna().sum() #Общ. кол-во
```

```python
df.isna().sum()/len(df) #Относительное кол-во
```

Вычисляю среднее для пропусков до полной обработки пропусков

```python
# Вычисляю среднее значение по группам для eu_sales
eu = df.groupby(['platform', 'year_of_release'])['eu_sales'].transform('mean')
# Заполняю пропуски
df['eu_sales'] = df['eu_sales'].fillna(eu)

# То же самое для jp_sales
jp = df.groupby(['platform', 'year_of_release'])['jp_sales'].transform('mean')
df['jp_sales'] = df['jp_sales'].fillna(jp)

print(df.info())
```

```python
#Исправляюсь и оставляю пропущенные значения в столбцах для дальнейшего анализа
df[['name', 'genre', 'rating']] = df[['name', 'genre', 'rating']].fillna('unknown') 
print(df.isna().sum())
```

### Выводы:  


- Пропуски в столбце с годом релиза также относительно не большое, однако удалять все 275 строк уже слишком, поэтому их оставим без изменения. Глобально на анализе это не скажется  
- Пропуски в столбцах с продажами скорее всего ошибка и их можно усреднить(т.к. столбец отвечает за регион, и как правило аудитория выставляет оценки приблизительно похожим играм одинаково, т.е. делаем ставку на вкусы определенной аудитории из региона)

## 3. Фильтрация данных

Коллеги хотят изучить историю продаж игр в начале XXI века, и их интересует период с 2000 по 2013 год включительно. Отберем данные по этому показателю. Сохраните новый срез данных в отдельном датафрейме, например `df_actual`.

```python
df_actual = df.query('year_of_release >= 2000 and year_of_release <= 2013').sort_values('year_of_release')
print(df_actual)
```

---

## 4. Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
def category_score (row):
    if 0 <= row['user_score'] < 3:
        return 'Низкая оценка'
    elif 3 <= row['user_score'] < 8:
        return 'Средняя оценка'
    elif 8 <= row['user_score'] <= 10:
        return 'Высокая оценка'
    elif pd.isna(row['user_score']):
        return 'Без оценки'
    
df_actual['user_score_category'] = df_actual.apply(category_score, axis = 1)
print(df_actual)
```

- Разделиим все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
def category_critic_score (row):
    if 0 <= row['critic_score'] < 30:
        return 'Низкая оценка'
    elif 30 <= row['critic_score'] < 80:
        return 'Средняя оценка'
    elif 80 <= row['critic_score'] <= 100:
        return 'Высокая оценка'
    elif pd.isna(row['critic_score']):
        return 'Без оценки'
    
df_actual['critic_score_category'] = df_actual.apply(category_critic_score, axis = 1)
print(df_actual)
```

- После категоризации данных проверим результат и посчитаем количество игр в каждой категории.

```python
count_user_score = df_actual['user_score_category'].value_counts()
print(count_user_score)
```

```python
count_critic_score = df_actual['critic_score_category'].value_counts()
print(count_critic_score)
```

- Выделим топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
top_platform = df_actual['platform'].value_counts()
print(top_platform.head(7))
```
